```python
\
BLUE = "\033[94m"
GREEN = "\033[92m"
RED = "\033[91m"
YELLOW = "\033[93m"
RESET = "\033[0m"

patients = []


def line():
    print(BLUE + "=" * 45 + RESET)


# Add Patient
def add_patient():
    line()
    print(YELLOW + "📌 Add New Patient" + RESET)
    pid = input("Enter Patient ID: ")
    name = input("Enter Patient Name: ")
    age = input("Enter Patient Age: ")
    disease = input("Enter Disease: ")

    patient = {"ID": pid, "Name": name, "Age": age, "Disease": disease}
    patients.append(patient)

    print(GREEN + "✔ Patient added successfully!" + RESET)


# View Patients
def view_patients():
    line()
    print(YELLOW + "📋 All Patients" + RESET)

    if not patients:
        print(RED + "No patients found." + RESET)
        return

    for p in patients:
        print(f"{BLUE}ID:{RESET} {p['ID']}  |  "
              f"{BLUE}Name:{RESET} {p['Name']}  |  "
              f"{BLUE}Age:{RESET} {p['Age']}  |  "
              f"{BLUE}Disease:{RESET} {p['Disease']}")


# Search Patient
def search_patient():
    line()
    print(YELLOW + "🔍 Search Patient" + RESET)
    pid = input("Enter Patient ID: ")

    for p in patients:
        if p["ID"] == pid:
            print(GREEN + "✔ Patient Found:" + RESET, p)
            return

    print(RED + "✘ Patient not found." + RESET)


# Update Patient
def update_patient():
    line()
    print(YELLOW + "✏ Update Patient" + RESET)
    pid = input("Enter Patient ID: ")

    for p in patients:
        if p["ID"] == pid:
            print("Current Data:", p)
            p["Name"] = input("Enter new name: ")
            p["Age"] = input("Enter new age: ")
            p["Disease"] = input("Enter new disease: ")
            print(GREEN + "✔ Patient updated successfully!" + RESET)
            return

    print(RED + "✘ Patient not found." + RESET)


# Delete Patient
def delete_patient():
    line()
    print(YELLOW + "🗑 Delete Patient" + RESET)
    pid = input("Enter Patient ID: ")

    for p in patients:
        if p["ID"] == pid:
            patients.remove(p)
            print(GREEN + "✔ Patient deleted successfully!" + RESET)
            return

    print(RED + "✘ Patient not found." + RESET)


# Main Menu
while True:
    line()
    print(BLUE + "🏥 Hospital Management System" + RESET)
    print("1. Add Patient")
    print("2. View Patients")
    print("3. Search Patient")
    print("4. Update Patient")
    print("5. Delete Patient")
    print("6. Exit")

    choice = input(YELLOW + "Enter your choice (1-6): " + RESET)

    if choice == "1":
        add_patient()
    elif choice == "2":
        view_patients()
    elif choice == "3":
        search_patient()
    elif choice == "4":
        update_patient()
    elif choice == "5":
        delete_patient()
    elif choice == "6":
        print(GREEN + "Exiting program... Goodbye!" + RESET)
        break
    else:
        print(RED + "Invalid choice, try again." + RESET)
```
